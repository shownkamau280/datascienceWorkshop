VERTICAL BARCHART

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
```

```python
#LOADING THE DATASET
df =pd.read_csv(r"C:\Users\ADMIN\OneDrive\Desktop\WORKSHOP\dataset\retail_sales_customer_analytics_dataset.csv")
df.head(5)
```

```python
df.info()
```

```python
quantity_count=df['Quantity'].value_counts()
```

```python
#DRAWING A CHART
plt.bar(
    quantity_count.index,
    quantity_count.values,
    color=['#001133','#e600e6','#4d004d','#ffccff','#e6eeff','#0044cc','#6699ff']
    )
plt.title(
    'Quantity Bar Graph',
    color='#3366cc',
    fontweight='bold',
    fontstyle='italic',
    fontsize=20,
    fontfamily='verdana'
    )

plt.ylabel('Amount of Quantity')
plt.xlabel('No of Quantity')
plt.figtext(0.99,0.65,'Fig : 1',ha='right',rotation=90)

colors=[
    Patch(facecolor='#001133',label='3.0'),
    Patch(facecolor='#e600e6',label='5.0'),
    Patch(facecolor='#4d004d',label='1.0'),
    Patch(facecolor='#ffccff',label='20.0'),
    Patch(facecolor='#e6eeff',label='-2.0'),
    Patch(facecolor='#0044cc',label='2.0'),
    Patch(facecolor='#6699ff',label='10.0')
]

plt.Legend(
    handles=colors,
    title='QUANTITY',
    LOC='upper right'
)
```
